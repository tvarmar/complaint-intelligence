**Source:** Consumer Financial Protection Bureau (CFPB)

**Original files:**
- `CCDB_Export_15_January_2026_through_February_2026.csv`
- `CCDB_Export_16_March_2026.csv`

### Objectives

- Validate dataset structure and temporal coverage
- Assess narrative availability and quality
- Analyze Product and Issue distributions
- Detect duplicates and inconsistent labels
- Identify potential data leakage risks
- Define a justified dataset preparation strategy

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

JAN_FEB_PATH = RAW_DATA_DIR / "cfpb_jan_feb_2026.zip"
MARCH_PATH = RAW_DATA_DIR / "cfpb_march_2026.zip"

print("Project root:", PROJECT_ROOT)
print("January-February exists:", JAN_FEB_PATH.is_file())
print("March exists:", MARCH_PATH.is_file())
```

```python
import zipfile
import pandas as pd

COLUMNS = [
  "Complaint ID",
  "Date received",
  "Product",
  "Issue",
  "Consumer complaint narrative",
]


def load_cfpb_archive(zip_path: Path) -> pd.DataFrame:
  """Load selected columns from a CFPB ZIP archive."""

  with zipfile.ZipFile(zip_path) as archive:
    csv_files = [
      name for name in archive.namelist()
      if name.lower().endswith(".csv")
    ]

    if len(csv_files) != 1:
      raise ValueError(
        f"Expected one CSV file in {zip_path.name},"
        f"found {len(csv_files)}."
      )

    with archive.open(csv_files[0]) as csv_file:
      df = pd.read_csv(
        csv_file,
        usecols=COLUMNS,
        low_memory=False,
      )

  return df
```

```python
df_jan_feb = load_cfpb_archive(JAN_FEB_PATH)
df_march = load_cfpb_archive(MARCH_PATH)

print("January-February:", df_jan_feb.shape)
print("March:", df_march.shape)
```
